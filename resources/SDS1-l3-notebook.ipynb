## Data

Seks korte taler fra tre partier og to årgange. Læg mærke til formen: det er en
**liste** af **dictionaries**. Det skal ligne den form data har, når det kommer fra en API
eller en JSON-fil.


````python
import pandas as pd

tekster = [
    {"id": 1, "parti": "Parti A", "aar": 2022,
     "tekst": "Vi vil styrke velfærden og investere i uddannelse. Uddannelse er nøglen til lighed."},
    {"id": 2, "parti": "Parti B", "aar": 2022,
     "tekst": "Skatten skal ned. Vi vil investere i vækst og nye arbejdspladser."},
    {"id": 3, "parti": "Parti C", "aar": 2022,
     "tekst": "Klimaet kan ikke vente. Vi vil investere i grøn omstilling nu."},
    {"id": 4, "parti": "Parti A", "aar": 2023,
     "tekst": "Velfærden er under pres. Vi vil investere mere i uddannelse og i sundhed."},
    {"id": 5, "parti": "Parti B", "aar": 2023,
     "tekst": "Lavere skat giver vækst. Vi vil investere i erhvervslivet."},
    {"id": 6, "parti": "Parti C", "aar": 2023,
     "tekst": "Grøn omstilling skaber arbejdspladser. Vi vil investere i vedvarende energi og natur."},
]

print(len(tekster))
print(tekster[0]["parti"])
print(tekster[0]["tekst"])
````

Og her er `rens()` fra sidste øvelse. I skal ikke skrive den igen, men kør cellen,
så I har funktionen i hukommelsen.


````python
def rens(t):
    t = t.lower()
    for tegn in [".", ",", "!", "?"]:
        t = t.replace(tegn, "")
    return t.split()

print(rens("Er det HER? Ja, det er."))
````

### 2.2 Kobl med oplysninger om partierne

**Trin 1: hvad går ikke op?**

hvem findes kun i partitabellen?

To funktioner gør de to lister sammenlignelige:

**`set()`** smider gengangere væk. Kolonnen `df["parti"]` har seks værdier. Hvert
parti står der to gange, én gang per år. `set()` skærer det ned til de tre
partinavne, der faktisk findes. Det er samme funktion, I brugte til at tælle unikke
ord i lektion 2.

**`sorted()`** stiller dem i alfabetisk rækkefølge og giver en almindelig liste
tilbage. Et `set` har nemlig ingen fast rækkefølge, så uden `sorted()` kunne de to
linjer komme ud i hver sin orden og så er de svære at sammenligne med øjnene.


````python
partier = pd.DataFrame({
    "parti":    ["Parti A", "Parti B", "Parti D"],
    "blok":     ["rød", "blå", "blå"],
    "mandater": [45, 32, 12]
})

print(partier)
print()
print(sorted(set(df["parti"])))
print(sorted(set(partier["parti"])))
````
